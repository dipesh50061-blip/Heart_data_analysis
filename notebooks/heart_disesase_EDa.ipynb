# 3. Understanding Heart Disease

Heart disease refers to a group of conditions that affect the heart and the blood vessels supplying it. One of the most common forms is coronary artery disease (CAD), where the blood vessels that supply blood to the heart muscle become narrowed or blocked.

The heart needs a continuous supply of oxygen-rich blood to function properly. Over time, factors such as high blood pressure, high cholesterol, smoking, diabetes, obesity, physical inactivity, age, and family history can contribute to changes or damage in the blood vessels. A buildup of fatty substances called plaque inside the arteries can make them narrower and reduce blood flow to the heart.

If the blood supply to the heart becomes severely restricted or completely blocked, it can result in a heart attack. Heart disease can also involve problems with heart rhythm, heart valves, or the heart muscle itself.

## 3.1 Common Risk Factors

Some important factors associated with heart disease include:

- Age: The risk generally increases with age.
- High blood pressure: Puts additional stress on the heart and blood vessels.
- High cholesterol: Can contribute to plaque buildup in arteries.
- Smoking: Can damage blood vessels and increase cardiovascular risk.
- Diabetes: Can damage blood vessels over time.
- Obesity: Can increase the risk of several cardiovascular risk factors.
- Physical inactivity: Is associated with increased cardiovascular risk.
- Unhealthy diet: Excessive saturated fats, trans fats, salt, and calories can contribute to cardiovascular risk.
- Family history: Some cardiovascular risk is influenced by genetics.
- Lifestyle factors: Long-term unhealthy lifestyle patterns can contribute to cardiovascular risk.

## 3.2 Common Symptoms

Depending on the condition, symptoms may include:

- Chest pain or discomfort
- Shortness of breath
- Unusual tiredness
- Dizziness or light-headedness
- Rapid or irregular heartbeat
- Pain or discomfort that may spread to the arm, shoulder, neck, jaw, or back

However, some people may have heart disease without obvious symptoms, which is one reason why risk-factor assessment and medical evaluation are important.
